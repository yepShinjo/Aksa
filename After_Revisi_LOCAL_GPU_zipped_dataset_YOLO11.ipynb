```python
# =========================
# YOLOv11 Train + Metrics
# =========================
from ultralytics import YOLO
import os
import math

# ---- EDIT THESE PATHS ----
WEIGHTS = r"C:\Users\User\Downloads\yolo11s.pt"
DATA_YAML = r"G:\combined-dataset-modifiedNO-BOOK-jalan-jalan-rebalanced-70_15_15-stratified-augmented\data.yaml"
RUNS_DIR  = r"D:\SKRIPSI\runs"
RUN_NAME  = "combined-dataset-modifiedNO-BOOK-jalan-jalan-rebalanced-70_15_15-stratified-augmented-150e"
# --------------------------

# -------- Train --------
model = YOLO(WEIGHTS)

model.train(
    data=DATA_YAML,
    epochs=150,
    imgsz=640,
    batch=-1,          # auto-select largest batch that fits VRAM
    device=0,          # GPU 0
    workers=0,         # Windows-friendly
    optimizer="AdamW",
    amp=True,          # mixed precision (faster on NVIDIA)
    cos_lr=True,       # cosine LR schedule
    patience=20,       # early stop
    cache=True,        # cache images in RAM
    project=RUNS_DIR,
    name=RUN_NAME,
)

# -------- Params (“neurons”) --------
# (people often say "neurons" but what’s useful is total parameters)
try:
    total_params = sum(p.numel() for p in model.model.parameters())
    trainable_params = sum(p.numel() for p in model.model.parameters() if p.requires_grad)
except Exception:
    # fallback: re-load best weights into a fresh model object if needed
    best_path = os.path.join(RUNS_DIR, RUN_NAME, "weights", "best.pt")
    model = YOLO(best_path)
    total_params = sum(p.numel() for p in model.model.parameters())
    trainable_params = sum(p.numel() for p in model.model.parameters() if p.requires_grad)

# -------- Extra evaluation on TEST split --------
# (You can change split="val" if you prefer)
metrics = model.val(
    data=DATA_YAML,
    split="test",
    device=0,
    imgsz=640,
    save_json=False,
    plots=False
)

# Try the modern attributes first, then fall back to results_dict
precision = recall = f1 = map50 = map5095 = None
try:
    # Ultralytics >= 8.2 style
    precision = float(getattr(metrics.box, "p"))
    recall    = float(getattr(metrics.box, "r"))
    map50     = float(getattr(metrics.box, "map50"))
    map5095   = float(getattr(metrics.box, "map"))
except Exception:
    # Older-style access via dict keys
    rd = getattr(metrics, "results_dict", {}) or {}
    precision = rd.get("metrics/precision(B)")
    recall    = rd.get("metrics/recall(B)")
    map50     = rd.get("metrics/mAP50(B)")
    map5095   = rd.get("metrics/mAP50-95(B)")

# Compute F1 if we have P and R
if precision is not None and recall is not None and (precision + recall) > 0:
    f1 = 2 * precision * recall / (precision + recall)

# -------- Pull final losses from results.csv --------
import pandas as pd

results_csv = os.path.join(RUNS_DIR, RUN_NAME, "results.csv")
final_losses = {}
if os.path.exists(results_csv):
    df = pd.read_csv(results_csv)
    last = df.iloc[-1]
    # Common YOLO columns (may vary slightly by version)
    # Train losses
    for col in ["train/box_loss", "train/cls_loss", "train/dfl_loss"]:
        if col in df.columns:
            final_losses[col] = float(last[col])
    # Val losses
    for col in ["val/box_loss", "val/cls_loss", "val/dfl_loss"]:
        if col in df.columns:
            final_losses[col] = float(last[col])
else:
    print(f"[WARN] Could not find results.csv at: {results_csv}")

# --------- Pretty print summary ---------
def fmt(x):
    if x is None or (isinstance(x, float) and (math.isnan(x) or math.isinf(x))):
        return "n/a"
    return f"{x:.4f}" if isinstance(x, float) else str(x)

print("\n================ TRAINING SUMMARY ================")
print(f"Run dir: {os.path.join(RUNS_DIR, RUN_NAME)}")
print(f"Total params:     {total_params:,}")
print(f"Trainable params: {trainable_params:,}")
print(f"Precision (test): {fmt(precision)}")
print(f"Recall (test):    {fmt(recall)}")
print(f"F1 (test):        {fmt(f1)}")
print(f"mAP50 (test):     {fmt(map50)}")
print(f"mAP50-95 (test):  {fmt(map5095)}")
if final_losses:
    print("Final losses (last epoch in results.csv):")
    for k, v in final_losses.items():
        print(f"  - {k}: {v:.5f}")
else:
    print("Final losses: n/a (results.csv not found)")
print("==================================================\n")
```

```python
# ==============================
# Independent TEST evaluation
# ==============================
from ultralytics import YOLO
import os, math, csv

# ---- EDIT THESE ----
WEIGHTS   = r"D:\SKRIPSI\runs\combined-dataset-modifiedNO-BOOK-jalan-jalan-rebalanced-70_15_15-stratified-augmented-150e\weights\best.pt"  # or last.pt
DATA_YAML = r"G:\combined-dataset-modifiedNO-BOOK-jalan-jalan-rebalanced-70_15_15-stratified-augmented\data.yaml"
SAVE_CSV  = r"D:\SKRIPSI\runs\test_eval\combined-dataset-modifiedNO-BOOK-jalan-jalan-rebalanced-70_15_15-stratified-augmented-150e.csv"
DEVICE    = 0  # GPU index or 'cpu'
# --------------------

os.makedirs(os.path.dirname(SAVE_CSV), exist_ok=True)

model = YOLO(WEIGHTS)

# Run evaluation strictly on the test split (no training involved)
metrics = model.val(
    data=DATA_YAML,
    split="test",
    device=DEVICE,
    imgsz=640,
    save_json=False,
    plots=True,       # saves PR curves, confusion matrix, etc. in runs/val/
    verbose=False
)

# ---- Extract overall metrics (robust across v8/v11) ----
overall = {
    "precision": None,
    "recall": None,
    "f1": None,
    "map50": None,
    "map5095": None
}

try:
    # Newer Ultralytics objects
    overall["precision"] = float(getattr(metrics.box, "p"))
    overall["recall"]    = float(getattr(metrics.box, "r"))
    overall["map50"]     = float(getattr(metrics.box, "map50"))
    overall["map5095"]   = float(getattr(metrics.box, "map"))
except Exception:
    # Fallback to dict
    rd = getattr(metrics, "results_dict", {}) or {}
    overall["precision"] = rd.get("metrics/precision(B)")
    overall["recall"]    = rd.get("metrics/recall(B)")
    overall["map50"]     = rd.get("metrics/mAP50(B)")
    overall["map5095"]   = rd.get("metrics/mAP50-95(B)")

if overall["precision"] is not None and overall["recall"] is not None and \
   (overall["precision"] + overall["recall"]) > 0:
    overall["f1"] = 2 * overall["precision"] * overall["recall"] / (overall["precision"] + overall["recall"])

# ---- Per-class metrics (Precision, Recall, F1, AP50, AP50-95) ----
names = getattr(metrics, "names", None)  # dict: {class_id: name}
per_class_rows = []

# Try the common attributes first
p_cls = r_cls = ap50_cls = ap5095_cls = None
try:
    p_cls       = getattr(metrics.box, "p_class", None)
    r_cls       = getattr(metrics.box, "r_class", None)
    ap50_cls    = getattr(metrics.box, "ap50_class", None)
    ap5095_cls  = getattr(metrics.box, "ap_class", None)  # COCO mAP (0.50:0.95)
except Exception:
    pass

if p_cls is not None and r_cls is not None:
    for cid in range(len(p_cls)):
        p, r = float(p_cls[cid]), float(r_cls[cid])
        f1 = 2 * p * r / (p + r) if (p + r) > 0 else 0.0
        row = {
            "class_id": cid,
            "class_name": names.get(cid, str(cid)) if isinstance(names, dict) else str(cid),
            "precision": p,
            "recall": r,
            "f1": f1,
            "ap50": float(ap50_cls[cid]) if ap50_cls is not None else None,
            "ap50_95": float(ap5095_cls[cid]) if ap5095_cls is not None else None
        }
        per_class_rows.append(row)

# ---- Pretty print ----
def fmt(x):
    if x is None or (isinstance(x, float) and (math.isnan(x) or math.isinf(x))):
        return "n/a"
    return f"{x:.4f}"

print("\n========== TEST METRICS (overall) ==========")
print(f"Precision:  {fmt(overall['precision'])}")
print(f"Recall:     {fmt(overall['recall'])}")
print(f"F1:         {fmt(overall['f1'])}")
print(f"mAP@0.50:   {fmt(overall['map50'])}")
print(f"mAP@0.50:95 {fmt(overall['map5095'])}")
print("============================================\n")

if per_class_rows:
    print("Per-class metrics (first few):")
    for row in per_class_rows[:10]:
        print(f"- {row['class_name']}: P={row['precision']:.3f} R={row['recall']:.3f} F1={row['f1']:.3f} "
              f"AP50={fmt(row['ap50'])} AP50-95={fmt(row['ap50_95'])}")
    if len(per_class_rows) > 10:
        print(f"... and {len(per_class_rows)-10} more classes.")

# ---- Save CSV ----
with open(SAVE_CSV, "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    # Header
    w.writerow(["type","class_id","class_name","precision","recall","f1","map50","map50_95"])
    # Overall row
    w.writerow([
        "overall","-","ALL",
        overall["precision"], overall["recall"], overall["f1"],
        overall["map50"], overall["map5095"]
    ])
    # Per-class rows
    for r in per_class_rows:
        w.writerow([
            "class", r["class_id"], r["class_name"],
            r["precision"], r["recall"], r["f1"], r["ap50"], r["ap50_95"]
        ])

print(f"✅ Saved test metrics CSV to: {SAVE_CSV}")
print("📁 Plots (PR curves, confusion matrix) saved under the latest runs\\val folder.")
```
